```python
#selecionando colunas específicas
df.select(['total_bill','tip']).head()
```

```python
#Usando expressões (permite operações no select)
df.select([pl.col('total_bill'), (pl.col('tip') / pl.col('total_bill')*100).alias('percent_gorgeta')]).head()
```

```python
#filtrando dados declientes de deixaram uma gorgeta maior que 20%
df.filter(pl.col('tip')/pl.col('total_bill')*100 > 20).head()
```

```python
df.filter(pl.col('day') == 'Sun').head()
```

```python
df = df.with_columns([(pl.col('tip')/pl.col('total_bill')*100).alias('percent_gorgeta'),
                      (pl.col('total_bill') + pl.col('tip')).alias('total_pago')])
df.head()
```

```python
#Ordenando dados
df.sort('total_bill').head()
```

```python
df.sort('total_bill', descending=True).head()
```

```python
df.sort(['day', 'total_bill'], descending=[False, True]).head()
```

```python
#Média e total de gorjetas por dia
df.group_by('day').agg([
    pl.mean('tip').alias('media_gorjeta'),
    pl.sum('tip').alias('total_gorjeta'),
    pl.count().alias('qtd_registro')
])
```

```python
df.group_by(['day', 'sex']).agg([
    pl.mean('tip').alias('media_gorjeta'),
    pl.sum('tip').alias('total_gorjeta'),
    pl.count().alias('qtd_registro')
])
```

```python
#Soma de valores para colunas específicas
df.select([
    pl.col('total_bill'),
    pl.col('tip'),
    pl.sum_horizontal(['total_bill', 'tip']).alias('Soma_horizontal')
]).head()
```

```python
resultado = (
    df.filter(pl.col('day') == 'Sun')
    .group_by('sex')
    .agg([
        pl.mean('percent_gorgeta').alias('media_perc_gorjeta'),
        pl.sum('total_bill').alias('total_contas')
    ]).sort('media_perc_gorjeta', descending=True)
)
resultado
```
