```python
import pandas as pd
```

```python
#Criando um dataframe com pandas
df_pandas = pd.DataFrame({
    'produto': ['arroz', 'feijão', 'macarrão'],
    'quantidade': [50, 120, 130 ]
})
```

```python
df_pandas
```

```python
#Converter para polars
df_polars = pl.from_pandas(df_pandas)
df_polars
```

```python
display(df_polars)
```

```python
#Voltando para pandas
df_pandas_again = df_polars.to_pandas()
df_pandas_again
```
